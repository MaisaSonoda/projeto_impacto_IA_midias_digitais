# IA's mais adotadas pelas empresas

Importando bibliotecadas e lendo o dataframe

```python
import pandas as pd
import statistics
import matplotlib.pyplot as plt

df = pd.read_csv(r'C:\Users\User\Downloads\Global_AI_Content_Impact_Dataset.csv')
```

Aqui vemos uma divisão consideravelmente igualitária na base de dados. Contudo, pode-se inferir uma fatia maior dos usuários para a OpenAi por conta do DALL-E e do ChatGPT, especialmente hoje em dia, onde há integração de ambos.

```python
df_industry = df.groupby('Top AI Tools Used')['Top AI Tools Used'].count()
df_industry = df_industry.sort_values(ascending=False)

# Labels automáticos (ou use personalizados caso necessário)
labels = df_industry.index
sizes = df_industry.values
cmap = plt.cm.Set3 

# Plot
plt.figure(figsize=(6, 6))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140, colors = cmap(range(len(labels))))

plt.title('Distribuição de Uso das Ferramentas de IA', fontsize=14)
plt.tight_layout()
plt.show()
```

# Aumento médio na receita após a implementação das respectivas IA's

Aqui vemos a Synthesia liderando, com o ChatGPT e o Stable Difusion logo atrás.

```python
# Aumento na receita de cada IA
df_revenue = df.groupby('Top AI Tools Used')['Revenue Increase Due to AI (%)'].mean()
df_revenue = df_revenue.sort_values(ascending=False)

# Plotando o gráfico
plt.title('Aumento médio da receita por cada IA', fontsize=14)
plt.xlabel('Ferramentas de IA', fontsize=12)
plt.ylabel('Aumento da Receita (%)', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.bar(df_revenue.index, df_revenue.values)
plt.tight_layout()
plt.show()
```

# Confiança média em cada ferramenta pelos usuários

Um dado interessante a se ressaltar é que vemos a confiança na IA tendo resultados bem parecidos com o aumento em % da receita das empresas.

```python
# Confiança média
df_consumertrust = df.groupby('Top AI Tools Used')['Consumer Trust in AI (%)'].mean()
df_consumertrust = df_revenue.sort_values(ascending=False)
plt.bar(df_consumertrust.index, df_consumertrust.values)

plt.title('Confiança média em cada IA', fontsize=14)
plt.xlabel('Ferramentas de IA', fontsize=12)
plt.ylabel('Confiança média (%)', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

plt.show()
```

# Taxa média de adoção da IA no trabalho

Apesar dos resultados passados, pode-se perceber uma liderança relevante dos projetos da OpenAI. Nesse caso, o que se pode-se levantar como hipótese é a ideia de que modelos de chatbot e conversação como o ChatGPT podem influenciar em mais coisas no trabalho do que apenas imagens, e, somado com o forte nome da empresa no mercado, esses usuários usam tais ferramentas para mais situações no dia a dia, mesmo que outras ferramentas (como o MidJourney) tenham mais usuários na prática.

```python
df_adoptionrate = df.groupby('Top AI Tools Used')['AI Adoption Rate (%)'].mean()
df_adoptionrate = df_adoptionrate.sort_values(ascending=False)
plt.bar(df_adoptionrate.index, df_adoptionrate.values)

plt.title('Taxa média de adoção da IA no trabalho', fontsize=14)
plt.xlabel('Ferramentas de IA', fontsize=12)
plt.ylabel('Taxa média (%)', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

plt.show()
```
