# IAs mais adotadas pelas empresas
Aqui vemos uma divisão consideravelmente igualitária na base de dados. Contudo, pode-se inferir uma fatia maior dos usuários para a OpenAi por conta do DALL-E e do ChatGPT, especialmente hoje em dia, onde há integração de ambos.

```python
import pandas as pd
import matplotlib.pyplot as plt
import psycopg2
from dotenv import load_dotenv
import os

load_dotenv()
conn = psycopg2.connect(database=os.getenv("DB_NAME"), user=os.getenv("DB_USER"), password=os.getenv("DB_PASSWORD"))

df = pd.read_sql_query("""
    SELECT top_ai_tools_used, COUNT(*) AS quantidade
    FROM ai_metrics
	GROUP BY top_ai_tools_used
    ORDER BY quantidade DESC
""", conn)

labels = df['top_ai_tools_used']
cmap = plt.cm.Set3 

plt.figure(figsize=(8,8))
plt.pie(df['quantidade'], labels = df['top_ai_tools_used'], autopct = '%1.1f%%', startangle = 140, colors = cmap(range(len(labels))))
plt.title('Distribuição do uso de IAs nas indústrias', fontsize = 20)
plt.tight_layout()
plt.show()
```

# Aumento médio na receita após a implementação das respectivas IAs
Aqui vemos a Synthesia liderando, com o ChatGPT e o Stable Difusion logo atrás.

```python
df = pd.read_sql_query("""
    SELECT top_ai_tools_used, AVG(revenue_increase_due_to_ai) AS Aumento_receita
    FROM ai_metrics
	GROUP BY top_ai_tools_used
	ORDER BY Aumento_receita DESC
""", conn)

plt.figure(figsize=(10, 6))
plt.title('Aumento médio da receita por cada IA', fontsize=14)
plt.xlabel('Ferramentas de IA', fontsize=12)
plt.ylabel('Aumento da Receita (%)', fontsize=12)
plt.bar(df['top_ai_tools_used'], df['aumento_receita'])
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

# Confiança média em cada ferramenta pelos usuários
Um dado interessante a se ressaltar é que vemos a confiança na IA tendo resultados bem parecidos com o aumento em % da receita das empresas.

```python
df = pd.read_sql_query("""
    SELECT top_ai_tools_used, AVG(consumer_trust_in_ai) AS confiança_do_consumidor
    FROM ai_metrics
	GROUP BY top_ai_tools_used
	ORDER BY confiança_do_consumidor DESC
""", conn)

plt.figure(figsize=(10, 6))
plt.title('Confiança média dos consumidores em cada IA', fontsize=14)
plt.xlabel('Ferramentas de IA', fontsize=12)
plt.ylabel('Confiança média (%)', fontsize=12)
plt.bar(df['top_ai_tools_used'], df['confiança_do_consumidor'])
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

# Taxa média de adoção da IA no trabalho
Apesar dos resultados passados, pode-se perceber uma liderança relevante dos projetos da OpenAI. Nesse caso, o que se pode-se levantar como hipótese é a ideia de que modelos de chatbot e conversação como o ChatGPT podem influenciar em mais coisas no trabalho do que apenas imagens, e, somado com o forte nome da empresa no mercado, esses usuários usam tais ferramentas para mais situações no dia a dia, mesmo que outras ferramentas (como o MidJourney) tenham mais usuários na prática.

```python
df = pd.read_sql_query("""
    SELECT top_ai_tools_used, AVG(ai_adoption_rate) AS taxa_de_adocao_das_ias
    FROM ai_metrics
	GROUP BY top_ai_tools_used
	ORDER BY taxa_de_adocao_das_ias DESC
""", conn)

plt.figure(figsize=(10, 6))
plt.title('Taxa de adoção/uso média no dia a dia dos consumidores em cada IA', fontsize=14)
plt.xlabel('Ferramentas de IA', fontsize=12)
plt.ylabel('Taxa de uso (%)', fontsize=12)
plt.bar(df['top_ai_tools_used'], df['taxa_de_adocao_das_ias'])
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
